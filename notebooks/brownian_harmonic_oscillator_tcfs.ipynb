```python
from rpmdgle.utils import ou
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# Oscillator parameters
mass = 1.0                       # mass
omega = 1.0                      # harmonic frequency
beta = 1.0                       # 1/kB*T
# mass-weighted drift matrix, [time^(-1)];  1x1 for Markovian case   
Ap = np.array([[0.25*omega]])
```

```python
# compute the TCFs
t = np.linspace(0, 10, 500)
cxx = ou.gle_cxx(t, omega, Ap, beta=beta, mass=mass)
```

```python
cqq = cxx[...,0,0]
cqp = cxx[...,0,1]
cpp = cxx[...,1,1]
fig, (ax_qq, ax_qp, ax_pp) = plt.subplots(nrows=3, sharex=True, figsize=(4,5))
ax_qq.plot(t, cqq, label="position ATCF")
ax_qp.plot(t, cqp, label="position-momentum")
ax_pp.plot(t, cpp, label="momentum ATCF")
ax_pp.set_xlabel('time')
for ax in [ax_qq, ax_qp, ax_pp]:
    ax.legend()
```
